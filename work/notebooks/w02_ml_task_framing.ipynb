## 3) Success metric

**Success metric: Precision@K**

We therefore care most about whether the pages near the top of the ranking are genuinely useful refresh opportunities. K can represent the number of pages the team has capacity to review or refresh.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## 5) Why ML beats a fixed rule here

A fixed rule, such as prioritizing every page with a CTR below a particular threshold, would treat one metric as the main decision factor. However, refresh priority can depend on several signals at the same time, including search demand, impressions, clicks, CTR, position, and content freshness.

These signals can interact in ways that are difficult to capture with one manually chosen rule. An ML-based ranking approach could learn patterns across multiple signals and produce a more flexible prioritization of pages.

The output would support a content action: the team could review the highest-ranked pages first and decide which ones are worth refreshing.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
low_ctr = df[df["ctr"] < 1]

print("Pages with CTR < 1%:", len(low_ctr))

display(
    low_ctr[
        ["content_id", "search_volume", "impressions_90d", "clicks_90d", "ctr"]
    ]
    .sort_values("impressions_90d", ascending=False)
    .head(10)
)
```
