## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

More than half of the stock (54.2%) is decreasing, and a large part of that (72.2%) remains visible; an actual chance to capitalize. It is important to mention that the decline ratio does not go up in a consistent manner alongside staleness (it was equal to 61.1% for 91 to 180 days old inventory but then decreased after 181 days), meaning that applying a simple rule, “mark anything which is old,” would fail.


```python
import os, urllib.request
import pandas as pd

path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(path):
    os.makedirs("data/raw", exist_ok=True)
    url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
    urllib.request.urlretrieve(url, path)

df = pd.read_csv(path)

declining = df["trend_direction"] == "down"
print(f"Declining pages: {declining.sum()} of {len(df)} = {declining.mean()*100:.1f}%")

visible_declining = df[declining & (df["avg_position"] > 0) & (df["avg_position"] <= 20)]
print(f"Declining AND top-20 visible: {len(visible_declining)} = "
      f"{len(visible_declining)/declining.sum()*100:.1f}% of declining pages")

decline_rate_by_freshness = pd.crosstab(df["freshness_tier"], df["trend_direction"], normalize="index")["down"] * 100
print("\nDecline rate by freshness tier (%):")
print(decline_rate_by_freshness.round(1))
```
